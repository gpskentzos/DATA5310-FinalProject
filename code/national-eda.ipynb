## Data Summary
taken from the data prep notebook

1. Crime Data: `crime`

    - **Source**: The [Crime Index](crimeindex.org/download) by Jeff Asher and his firm AH Datalytics. Compiled from FBI Uniform Crime Reporting data

    - **Info**:
      - Downloaded November 2026
      - Contains the yearly crime counts and rates for police agencies across the U.S.
      - Each row is one location/place (city, county, state, or the whole nation) in one year.
      - 204,785 rows and 23 columns
      - Time range: 2010-2024

    - **Key Variables**:
      - `name`, `id` - identifiers for location
      - `population`
      - `X_total` - total # of crimes per year for that crime
      - `violent_capita` - murder, rape, robbery, agg assault crimes per 100k residents
      - `property_capita` - burglary, theft, motor vehicle theft per 100k residents
    
    - **Notes**:
      - We kept only cities
      - Empty columns were dropped
      - Entries with no population were also dropped
    


2. City Characteristics: `census`
   - **Source**: U.S. Census Bureau ACS 5-Year Data Profiles. Tables DP03 (economic) and DP04 (housing). From data.census.gov.

    - **Info**:
      - Downloaded November 2026
      - Contains economic and housing characteristics for every U.S. city and town.
      - Each row is one place in each 5-year period.
      - 91,452 rows, 4 columns (that we selected)
      - 3 time periods: 2010–2014, 2015–2019, 2020–2024

    - **Key Variables**:
      - `median_income` - median household income ($ U.S. dollars)
      - `pct_poverty` - % of people below the poverty line
      - `pct_uninsured` - % of people with no health insurance
      - `median_rent` - our measure for **cost of living**
      - `city`, and `state` - location
      - `period` - census time range

    - **Notes**:
      - We will be focusing on larger cities in our analysis
      - Some of the cities have odd naming and may cause issues later when joining to the crime data.

```python
# apply a filter so we can look at only cities
# we won't care about small towns, etc
larger = crime[crime['population'] >= 100000]   # skip tiny towns
```

### 1. What crime types are correlated to one another?
I am thinking I could start by analyzing what types of crimes appear to have a correlation. <br>
The best way I can think to show this would be a **correlation heatmap**.

```python
# To avoid error I kept getting,
# create a holder for the variables we want to use
rate_columns = ['murder_capita', 'rape_capita', 'robbery_capita', 'assault_capita',
             'burglary_capita', 'theft_capita', 'motor_capita']
```

-> Notice how we are using the `_capita` NOT the `_total`, <br>
This is because we want to use the count in relation to the location size. <br>
The data may looked innaccuraate if we don't.

```python
correlation = larger[rate_columns].corr().reset_index().melt(id_vars='index', var_name='var2', value_name='r')

base = alt.Chart(correlation).encode(x=alt.X('index:N', title=None), y=alt.Y('var2:N', title=None))
heat = base.mark_rect().encode(color=alt.Color('r:Q', scale=alt.Scale(scheme='redblue', domain=[-1, 1], reverse=True)))
text = base.mark_text(fontSize=10).encode(text=alt.Text('r:Q', format='.2f'))

(heat + text).properties(title='Correlation between crime types (cities over 100k)', width=450, height=450)
```

### 2. Do bigger cities have more crime?
I want to see how **city size vs. violent crime**

```python
# using the same 'larger' filter we made before
alt.Chart(larger[larger['year'] == 2024]).mark_circle(opacity=0.5).encode(
    x=alt.X('population:Q', scale=alt.Scale(type='log'), title='Population (log scale)'),
    y=alt.Y('violent_capita:Q', title='Violent crimes per 100k'),
).properties(title='City size vs. violent crime rate (2024)', width=450)
```

Do the same but add in an **interaction** for practice:

```python
# using the same 'larger' filter we made before
# incorporating tooltip interaction - topics we are covering this week
alt.Chart(larger[larger['year'] == 2024]).mark_circle(opacity=0.5).encode(
    x=alt.X('population:Q', scale=alt.Scale(type='log'), title='Population (log scale)'),
    y=alt.Y('violent_capita:Q', title='Violent crimes per 100k'),
    tooltip=['name', 'population', 'violent_capita']
).properties(title='City size vs. violent crime rate (2024)', width=450)
```

### 3. Do cities with higher property crime also have higher violent crime?

```python
alt.Chart(larger[larger['year'] == 2024]).mark_circle(opacity=0.5).encode(
    x=alt.X('property_capita:Q', title='Peroperty crimes per 100k'),
    y=alt.Y('violent_capita:Q', title='Violent crimes per 100k'),
    color=alt.condition(alt.datum.id == 'WASPD0000', alt.value('#d62728'), alt.value('#888')),
    size=alt.condition(alt.datum.id == 'WASPD0000', alt.value(150), alt.value(30)),
    tooltip=['name', 'violent_capita', 'property_capita'] # interaction
).properties(title='Property vs. Violent crime (2024), Seattle in red', width=450)
```

I want to look at the inverse of the above

```python
alt.Chart(larger[larger['year'] == 2024]).mark_circle(opacity=0.5).encode(
    x=alt.X('violent_capita:Q', title='Violent crimes per 100k'),
    y=alt.Y('property_capita:Q', title='Property crimes per 100k'),
    color=alt.condition(alt.datum.id == 'WASPD0000', alt.value('#d62728'), alt.value('#888')),
    size=alt.condition(alt.datum.id == 'WASPD0000', alt.value(150), alt.value(30)),
    tooltip=['name', 'violent_capita', 'property_capita'] # interaction
).properties(title='Violent vs. Property crime (2024), Seattle in red', width=450)
```

#### Summary

From the visuals above, I can see that:
- There is a clear correlation between increases in both violent and property crime. <br>
Meaning that when one increases, so does the other - no matter how we change the x/y positioning.
- When the y-position axis was changed to **violent crime**, there is a more sharp incline in property crime
  - This tells me that there is a correlation between property and vriolent crime <br>
  BUT more violent crime creates a more abrupt increase in proeperty crime.
  - Property crime does not create as sharp of an increase in violent crime.
- Seattle sits on the lower-end of violent crime but on the higher-end of property crime.
- Memphis is another outlier for both property and violent crime.

### 4. Do cities with higher poverty rates have higher crime? Violent or Property?
I am envisioning a faceted display where each chart shows each economic variable in relation to one another

```python
# focus on recent census data for now
recent_data = census[census['period'] == '2020-2024'].dropna()
# choose economic variables to focus on
ec_vars = ['median_income', 'pct_poverty', 'pct_uninsured', 'median_rent']

alt.Chart(recent_data.sample(3000, random_state=1)).mark_circle(size=8, opacity=0.3).encode(
    x=alt.X(alt.repeat('column'), type='quantitative'),
    y=alt.Y(alt.repeat('row'), type='quantitative')
).properties(width=150, height=150).repeat(row=ec_vars, column=ec_vars)
```

#### Summary
Each row is 1 of the 4 economic variables we chose to focus on. <br>
Looking at the x/y-positioning, we can see how each variable relates to another.
- When looking at `median_income` on the x-axis, we can see that there is a **negative relationship** between median income and amount of residents uninsured.
  - meaning that as median income increases, more people seem to have health insurance coverage.
- We can see that **as median income increases, poverty levels decrease**
- What is interesting to me.. <br>
It appears that **as median rent increases, poverty levels decrease**. <br>
